```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('data01.csv')

medianHouse = df['median_house_value'].median()
print(f'Median House Value: {medianHouse}')
for x in df.index:
  if df.loc[x, 'total_rooms'] < 500:
    df.loc[x, 'total_rooms'] = int(df['total_rooms'].median())
for x in df.index:
  if df.loc[x, 'housing_median_age'] < 25:
    print(f'{df.loc[x]}\n')

df.plot(kind = 'scatter', x = 'median_house_value', y = 'total_bedrooms')
plt.show()

df

```
